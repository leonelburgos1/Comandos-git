```python
import cv2 #Computer Vision Two
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

img = cv2.imread(file_name, cv2.IMREAD_GRAYSCALE)

print("Original Image Size: (height, width)", img.shape)
plt.figure(figsize=(5,5))
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')
plt.show()

widht = img.shape[1] * 3
height = img.shape[0]*3

img_nearest = cv2.resize(img, (widht, height), interpolation=cv2.INTER_NEAREST)
img_nearest_e = cv2.resize(img, (widht, height), interpolation=cv2.INTER_NEAREST_EXACT)

img_linear = cv2.resize(img, (widht, height), interpolation=cv2.INTER_LINEAR)
img_linear_e = cv2.resize(img, (widht, height), interpolation=cv2.INTER_LINEAR_EXACT)

img_cubic = cv2.resize(img, (widht, height), interpolation=cv2.INTER_CUBIC)

fig, axis = plt.subplots(1, 3, figsize=(16, 6))

axis[0].imshow(img_nearest, cmap='gray')
axis[0].set_title('Nearest Interpolation')
axis[0].axis('off')

axis[1].imshow(img_linear, cmap='gray')
axis[1].set_title('Linear Interpolation')
axis[1].axis('off')

axis[2].imshow(img_cubic, cmap='gray')
axis[2].set_title('Cubic Interpolation')
axis[2].axis('off')

plt.tight_layout()
plt.show()

zoom = 250

fig, axis = plt.subplots(2, 3, figsize= (18, 6))

axis[0, 0].imshow(img_nearest[0:zoom, 0:zoom], cmap='gray')
axis[0, 0].set_title('Nearest Interpolation - Zoom')
axis[0, 0].axis('off')

axis[1, 0].imshow(img_nearest_e[0:zoom, 0:zoom], cmap='gray')
axis[1, 0].set_title('Nearest Interpolation E - Zoom')
axis[1, 0].axis('off')

axis[0, 1].imshow(img_linear[0:zoom, 0:zoom], cmap='gray')
axis[0, 1].set_title('Lineal Interpolation - Zoom')
axis[0, 1].axis('off')

axis[1, 1].imshow(img_linear_e[0:zoom, 0:zoom], cmap='gray')
axis[1, 1].set_title('Lineal Interpolation E - Zoom')
axis[1, 1].axis('off')

axis[0, 2].imshow(img_cubic[0:zoom, 0:zoom], cmap='gray')
axis[0, 2].set_title('Cubic Interpolation - Zoom')
axis[0, 2].axis('off')

plt.tight_layout()
plt.show()
```
